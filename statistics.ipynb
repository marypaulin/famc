# Test data statistics

```python
from matplotlib import pyplot as plt

import config
import loader
import counter
```

## Statistics for laat data

```python
data, train_data, valid_data, test_data, vocab, laat_args_new = loader.load_laat_data(config.LAAT_ARGS)
test_dataloader = loader.create_laat_dataloader(test_data, vocab, laat_args_new)
```

### Whole test set

```python
n_samples, n_tokens, n_labels = counter.count_all(test_dataloader)
print("Number of samples:", n_samples)
```

```python
# Plot histogram of numbers of tokens
n, bins, patches = plt.hist(n_tokens, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of tokens')
plt.ylabel('Probability')
plt.title('Histogram of numbers of tokens')
plt.xlim(min(n_tokens), max(n_tokens))
plt.ylim(0, 0.001)
plt.grid(True)
plt.show()
```

```python
# Plot histogram of numbers of labels
n, bins, patches = plt.hist(n_labels, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of labels')
plt.ylabel('Probability')
plt.title('Histogram of numbers of labels')
plt.xlim(min(n_labels), max(n_labels))
plt.ylim(0, 0.1)
plt.grid(True)
plt.show()
```

### Subset used for experiments

```python
n_subsamples, n_subtokens, n_sublabels = counter.count_sub(test_dataloader)
print("Number of samples in subset:", n_subsamples)
```

```python
# Plot histogram of numbers of tokens in subset
n, bins, patches = plt.hist(n_subtokens, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of tokens')
plt.ylabel('Probability')
plt.title('Histogram of numbers of tokens in subset')
plt.xlim(min(n_subtokens), max(n_subtokens))
plt.ylim(0, 0.002)
plt.grid(True)
plt.show()
```

```python
# Plot histogram of numbers of labels
n, bins, patches = plt.hist(n_sublabels, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of labels')
plt.ylabel('Probability')
plt.title('Histogram of numbers of labels in subset')
plt.xlim(min(n_sublabels), max(n_sublabels))
plt.ylim(0, 0.2)
plt.grid(True)
plt.show()
```
