# Test data statistics

```python
from matplotlib import pyplot as plt
import numpy as np
import torch

import config
import loader
```

## Statistics for laat data

```python
laat_args = config.LAAT_ARGS
data, train_data, valid_data, test_data, vocab, laat_args_new = loader.load_laat_data(laat_args)
```

### Whole test set

```python
test_dataloader = loader.create_laat_dataloader(test_data, vocab, laat_args_new)
```

```python
n_samples = 0
n_tokens = []
n_labels = []
for idx, tup in enumerate(test_dataloader):
    input_indices, labels, afa, _ = tup
    # Count samples
    n_samples += 1
    # Count tokens per sample
    non_zero_indices = torch.nonzero(input_indices[0], as_tuple=False)
    n_token = non_zero_indices.size()[0]
    n_tokens.append(n_token)
    # Count labels per sample
    non_zero_labels = torch.nonzero(labels[1][0], as_tuple=False)
    n_label = non_zero_labels.size()[0]
    n_labels.append(n_label)

min_tokens = min(n_tokens)
max_tokens = max(n_tokens)
min_labels = min(n_labels)
max_labels = max(n_labels)
print("Number of samples:", n_samples)
print(f"Min/max number of tokens per sample: {min_tokens}, {max_tokens}")
print(f"Min/max number of true labels per sample: {min_labels}, {max_labels}")
```

```python
# Plot histogram of numbers of tokens
n, bins, patches = plt.hist(n_tokens, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of tokens')
plt.ylabel('Probability')
plt.title('Histogram of numbers of tokens')
plt.xlim(min_tokens, max_tokens)
plt.ylim(0, 0.001)
plt.grid(True)
plt.show()
```

```python
# Plot histogram of numbers of labels
n, bins, patches = plt.hist(n_labels, 50, density=True, facecolor='b', alpha=0.75)
plt.xlabel('Numbers of labels')
plt.ylabel('Probability')
plt.title('Histogram of numbers of labels')
plt.xlim(min_labels, max_labels)
plt.ylim(0, 0.1)
plt.grid(True)
plt.show()
```
