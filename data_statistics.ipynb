# Test data statistics

```python
from matplotlib import pyplot as plt
import numpy as np

import config
import loader
import counter
```

## LAAT

```python
# Load laat data
test_data, vocab, laat_args_new = loader.load_laat_data(config.LAAT_ARGS)
laat_dataloader = loader.create_laat_dataloader(test_data, vocab, laat_args_new)
```

```python
# Get stats for laat data
n_samples, n_tokens, n_labels, n_subsamples, n_subtokens, n_sublabels = counter.count_samples_tokens_labels('laat', laat_dataloader)
freq, freq_sub = counter.count_label_frequencies('laat', laat_dataloader)
print("Number of samples in whole set:", n_samples)
print("Number of samples in subset:", n_subsamples)
print("Number of labels in whole set:", len(freq))
print("Number of labels in subset:", np.count_nonzero(freq_sub))
```

```python
# Plot stats for laat data
fig, axs = plt.subplots(2, 3, figsize=(12,5))

# Plot n_tokens
n_bins = 50
# Whole set
axs[0, 0].hist(n_tokens, n_bins, density=True, facecolor='b', alpha=0.75)
axs[0, 0].set_title("# Tokens per sample in whole set")
axs[0, 0].set_ylabel("Probability")
axs[0, 0].tick_params(labelbottom=False)
# Subset
axs[1, 0].hist(n_subtokens, n_bins, density=True, facecolor='b', alpha=0.75)
axs[1, 0].sharex(axs[0, 0])
axs[1, 0].set_title("# Tokens per sample in subset")
axs[1, 0].set_xlabel("# Tokens per sample")
axs[1, 0].set_ylabel("Probability")

# Plot n_labels
# Whole set
axs[0, 1].hist(n_labels, n_bins, density=True, facecolor='b', alpha=0.75)
axs[0, 1].set_title("# Labels per sample in whole set")
axs[0, 1].tick_params(labelbottom=False)
# Subset
axs[1, 1].hist(n_sublabels, n_bins, density=True, facecolor='b', alpha=0.75)
axs[1, 1].sharex(axs[0, 1])
axs[1, 1].set_title("# Labels per sample in subset")
axs[1, 1].set_xlabel("# Labels per sample")

# Plot label frequencies
bin_size = 50
n_bins = len(freq)//bin_size
# Whole set
axs[0, 2].hist(np.arange(len(freq)), n_bins, weights=freq, density=True, facecolor='b', alpha=0.75)
axs[0, 2].set_title("Label frequencies in whole set")
axs[0, 2].tick_params(labelbottom=False)
# Subset
axs[1, 2].hist(np.arange(len(freq)), n_bins, weights=freq_sub, density=True, facecolor='b', alpha=0.75)
axs[1, 2].sharex(axs[0, 2])
axs[1, 2].set_title("Label frequencies in subset")
axs[1, 2].set_xlabel(f"Label bins (ordered by frequency)")

fig.tight_layout()
```

```python
# Save stats for laat data
fig.savefig(f"stats/laat_nsub_{config.N_SUB}_seed_{config.SEED}.eps", format="eps")
```
