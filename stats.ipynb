# Test data statistics

```python
from matplotlib import pyplot as plt

import config
import loader
import counter
```

## Statistics for laat data

```python
data, train_data, valid_data, test_data, vocab, laat_args_new = loader.load_laat_data(config.LAAT_ARGS)
test_dataloader = loader.create_laat_dataloader(test_data, vocab, laat_args_new)
```

```python
n_samples, n_tokens, n_labels = counter.count_all(test_dataloader)
n_subsamples, n_subtokens, n_sublabels = counter.count_sub(test_dataloader)
print("Number of samples in whole set:", n_samples)
print("Number of samples in subset:", n_subsamples)
```

```python
# Plot stats for laat
fig, axs = plt.subplots(2, 2, figsize=(7,5))
axs[0, 0].hist(n_tokens, 50, density=True, facecolor='b', alpha=0.75)
axs[0, 0].set_title("Numbers of tokens in whole set")
axs[0, 0].set_ylabel("Probability")
axs[1, 0].hist(n_subtokens, 50, density=True, facecolor='b', alpha=0.75)
axs[1, 0].sharex(axs[0, 0])
axs[1, 0].set_title("Numbers of tokens in subset")
axs[1, 0].set_xlabel("Numbers of tokens")
axs[1, 0].set_ylabel("Probability")
axs[0, 1].hist(n_labels, 50, density=True, facecolor='b', alpha=0.75)
axs[0, 1].set_title("Numbers of labels in whole set")
axs[1, 1].hist(n_sublabels, 50, density=True, facecolor='b', alpha=0.75)
axs[1, 1].sharex(axs[0, 1])
axs[1, 1].set_title("Numbers of labels in subset")
axs[1, 1].set_xlabel("Numbers of labels")
for ax in fig.get_axes():
    ax.label_outer()

fig.tight_layout()
```

```python
fig.savefig(f"stats/laat_nsub_{config.N_SUB}_seed_{config.SEED}.eps", format="eps")
```
