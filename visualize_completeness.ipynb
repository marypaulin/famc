# Visualize completeness results

Run check_completeness.py first.

```python
from matplotlib import pyplot as plt

import attributor
import config
```

## Read results from files

```python
results = {}
method_name = 'ig'
for model_name in config.MODELS:
    results[model_name] = {}
    for n_steps in [10, 100, 200, 300]:
        method_name_new = str(n_steps)
        df = attributor.read_completeness_from_file(model_name,
                                                    method_name,
                                                    n_steps)
        if df is not None:
            results[model_name][method_name_new] = df                    
```

```python
print(results['caml'].keys())
```

# Plot results

```python
for model_name in config.MODELS:
    fig, axs = plt.subplots(1, 4, figsize=(15,4))
    for i, method_name in enumerate(results[model_name].keys()):
        xlabel = f"sum(attrs) for n_steps={method_name}"
        ylabel = 'pred_input - pred_base'
        sums = results[model_name][method_name]['sums'].values
        diffs = results[model_name][method_name]['diffs'].values
        # Scatterplot of sums and diffs
        axs[i].scatter(sums, diffs)
        if model_name == 'caml':
            axs[i].set_xlim(0.0,1.0)
            axs[i].set_ylim(0.0,1.0)
        elif model_name == 'laat':
            axs[i].set_xlim(-0.5,1.5)
            axs[i].set_ylim(-0.5,1.5)
        axs[i].set_xlabel(xlabel, fontsize=13)
        if i == 0:
            axs[i].set_ylabel(ylabel, fontsize=13)
    fig.tight_layout()
    # Save plot
    basename = f"plots/scatterplots_completeness_{model_name}"
    fig.savefig(f"{basename}.eps", format="eps")
    fig.savefig(f"{basename}.png", format="png")
```
